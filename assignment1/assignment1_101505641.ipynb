```python
# Author: Gurnoor Khurana
# Assignment #1

# Defining Variables
gym_member = "Jordan Michaels"  # String
target_weight_kg = 22.7  # Float
max_reps = 30  # Integer
is_active_member = False  # Boolean

# Dictionary storing workout stats; Structure: {Name: (Yoga, Running, Weightlifting)}
workout_stats = {
    "Sam": (35, 40, 20),
    "Taylor": (45, 35, 30),
    "Chris": (28, 50, 18)
}

# Compute total workout minutes and update dictionary
friend_names = list(workout_stats.keys())  # Store original keys
for name in friend_names:
    total_time = sum(workout_stats[name])
    workout_stats[f"{name}_Total"] = total_time

# Convert workout data into a 2D list
data_matrix = [list(values) for values in workout_stats.values() if isinstance(values, tuple)]

# Extract Yoga and Running times for all friends
yoga_running_times = [entry[:2] for entry in data_matrix]
print("Yoga & Running Times:", yoga_running_times)

# Extract Weightlifting minutes for the last two friends
lifting_times = [entry[2] for entry in data_matrix[-2:]]
print("Weightlifting Minutes (Last Two):", lifting_times)

# Check and congratulate friends with total workout time >= 120
for person, minutes in workout_stats.items():
    if isinstance(minutes, int) and minutes >= 120:
        print(f"Awesome work, {person.split('_')[0]}! Keep it up!")

# User input to check specific friend's stats
friend_input = input("Enter a friend's name: ")

if friend_input in workout_stats:
    activity_stats = workout_stats[friend_input]
    if isinstance(activity_stats, tuple):
        print(f"{friend_input}'s workout: Yoga-{activity_stats[0]}, Running-{activity_stats[1]}, Weightlifting-{activity_stats[2]}")
        print(f"Total Time: {sum(activity_stats)} minutes")
else:
    print(f"No records found for {friend_input}.")

# Identify the most and least active friends
total_times = {key: value for key, value in workout_stats.items() if isinstance(value, int)}

most_active = max(total_times, key=total_times.get)
least_active = min(total_times, key=total_times.get)

print(f"Most Active: {most_active.split('_')[0]}")
print(f"Least Active: {least_active.split('_')[0]}")
```
